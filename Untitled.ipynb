```python

import os, librosa, numpy as np, torch
import torch.nn as nn
from sklearn.metrics import roc_auc_score
from torch.utils.data import Dataset, DataLoader

# 路径完全没变
normal_dir = '/home/jovyan/work/shared_data/MIMII/6_dB/pump/id_00/mini_data/normal'
abnormal_dir = '/home/jovyan/work/shared_data/MIMII/6_dB/pump/id_00/mini_data/abnormal'

def wav_to_logmel(path, sr=16000, n_mels=64, frames=32):
    y, _ = librosa.load(path, sr=sr)
    mel = librosa.feature.melspectrogram(y=y, sr=sr, n_mels=n_mels, n_fft=1024, hop_length=512)
    logmel = librosa.power_to_db(mel, ref=np.max)
    if logmel.shape[1] < frames:
        logmel = np.pad(logmel, ((0,0),(0, frames - logmel.shape[1])))
    else:
        logmel = logmel[:, :frames]
    return logmel.flatten()

class MIMIIDataset(Dataset):
    def __init__(self, normal_dir, abnormal_dir=None):
        self.files, self.labels = [], []
        for f in os.listdir(normal_dir):
            # ⚠️ 这里改成了 .flac！！！
            if f.lower().endswith('.flac'): 
                self.files.append(os.path.join(normal_dir, f)); self.labels.append(0)
        if abnormal_dir:
            for f in os.listdir(abnormal_dir):
                # ⚠️ 这里改成了 .flac！！！
                if f.lower().endswith('.flac'): 
                    self.files.append(os.path.join(abnormal_dir, f)); self.labels.append(1)
                    
    def __len__(self): return len(self.files)
    def __getitem__(self, idx):
        return torch.tensor(wav_to_logmel(self.files[idx]), dtype=torch.float32), torch.tensor(self.labels[idx], dtype=torch.long)

class AutoEncoder(nn.Module):
    def __init__(self, dim=2048):
        super().__init__()
        self.encoder = nn.Sequential(nn.Linear(dim, 256), nn.ReLU(), nn.Linear(256, 64), nn.ReLU(), nn.Linear(64, 8))
        self.decoder = nn.Sequential(nn.Linear(8, 64), nn.ReLU(), nn.Linear(64, 256), nn.ReLU(), nn.Linear(256, dim), nn.Sigmoid())
    def forward(self, x): return self.decoder(self.encoder(x))

device = torch.device('cpu') 

dataset = MIMIIDataset(normal_dir, abnormal_dir)
print(f"✅ 成功读取到 {len(dataset)} 个音频文件！准备开始训练...")

if len(dataset) == 0:
    print("❌ 错误：依然没找到文件！")
else:
    train_loader = DataLoader(dataset, batch_size=8, shuffle=True)
    model = AutoEncoder().to(device)
    optimizer = torch.optim.Adam(model.parameters(), lr=1e-3)
    criterion = nn.MSELoss()

    for epoch in range(5): 
        model.train()
        for x, _ in train_loader:
            x = x.to(device)
            loss = criterion(model(x), x)
            optimizer.zero_grad(); loss.backward(); optimizer.step()

    model.eval()
    scores, labels = [], []
    test_loader = DataLoader(dataset, batch_size=8)
    with torch.no_grad():
        for x, y in test_loader:
            x = x.to(device)
            err = torch.mean((model(x) - x) ** 2, dim=1) 
            scores.extend(err.cpu().numpy()); labels.extend(y.numpy())

    print(f"\n🎉 最终测试集 AUC: {roc_auc_score(labels, scores):.4f}")
```

```python
torch.save(model.state_dict(), 'model.pkl')
print("✅ 权重保存成功！")
```
